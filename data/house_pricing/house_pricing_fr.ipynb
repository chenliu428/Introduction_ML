# Apprentissage automatique (Machine Learning) pour la prédiction des prix de l’immobilier

## Description du projet

Vous avez été engagé en tant que consultant en machine learning par une agence immobilière afin de développer un modèle capable de prédire les prix des biens immobiliers à partir de différentes caractéristiques. La seule ressource mise à votre disposition est un jeu de données, `data.csv`, qui contient toutes les informations nécessaires pour mener à bien cette mission.

Le livrable final doit être un modèle de machine learning entièrement spécifié, comprenant :

* le format des données d’entrée,
* le format des données de sortie,
* l’architecture du modèle,
* les paramètres optimisés du modèle,

ainsi qu’une estimation raisonnable des performances réelles du modèle en situation réelle.

En plus du modèle final, vous devez documenter l’ensemble du processus de développement ayant conduit à votre solution. Votre rapport doit inclure, sans s’y limiter :

* l’inspection et le nettoyage des données,
* l’analyse exploratoire des données,
* l’ingénierie des variables, avec une justification des techniques utilisées,
* la conception et l’entraînement du modèle,
* l’analyse du surapprentissage et l’optimisation des hyperparamètres,
* l’évaluation du modèle et l’analyse de ses performances.


### Exigences

* L'ensemble du projet doit être soumis sous la forme de ce notebook Jupyter (`.ipynb`).
* Tout le code doit être entièrement exécutable sur n'importe quel ordinateur ou dans une nouvelle session Google Colab après avoir modifié le chemin d'accès au fichier de données (`data.csv`).
* Seuls les packages et modules Python suivants sont autorisés :

  * `numpy`
  * `matplotlib`
  * `pandas.read_csv`
  * `pandas.DataFrame`
  * `sklearn.linear_model`
