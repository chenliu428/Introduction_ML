# Machine Learning for house price prediction

## Project Description

You have been hired as a machine learning consultant by a real estate agency to develop a model that predicts house prices based on a variety of real estate features. The only resource provided to you is a dataset, `data.csv`, which contains all the available information needed to complete this task.

The final deliverable should be a fully specified machine learning model, including:

* the input data format,
* the output data format,
* the model architecture,
* the optimized model parameters,

as well as a reasonable estimate of the model's real-world predictive performance.

In addition to the final model, you are required to document the entire development process that led to your solution. Your report should include, but is not limited to:

* data inspection and cleaning,
* exploratory data analysis,
* feature engineering, including a justification for the chosen techniques,
* model design and training,
* overfitting analysis and hyperparameter optimization,
* model evaluation and performance assessment.

## Requirements

* The entire project must be submitted as this Jupyter notebook (`.ipynb`) file.
* All code must be fully executable on an arbitrary computer or in a fresh Google Colab session after updating the path to the dataset (`data.csv`).
* Only the following Python packages and modules may be used:

  * `numpy`
  * `matplotlib`
  * `pandas.read_csv`
  * `pandas.DataFrame`
  * `sklearn.linear_model`
