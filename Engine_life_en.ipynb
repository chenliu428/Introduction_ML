## Background

Engines start with varying levels of initial wear and manufacturing differences, which are unknown to the user. Initially, engines operate under normal conditions and begin to degrade over time. This degradation progressively increases until a predefined threshold is reached, at which point the engine is considered unsafe for further use.

You are given multivariate data from aircraft engines of the same type, recorded at random operational cycles. The dataset includes three *setting* parameters, 21 sensor readings, and the corresponding *lifetime* at that state. The *lifetime*, obtained a posteriori, represents the remaining number of operational cycles before the degradation threshold is reached.

The objective is to predict the lifetime of this engine type based on the setting parameters and sensor readings at a given operational state, using linear regression models.

The data is split into a training dataset (`train_data.dat`) and a test dataset (`test_data.dat`). The test dataset is exclusively reserved for evaluating the performance of the trained model on unseen data, and model training must be performed exclusively using the training dataset.
