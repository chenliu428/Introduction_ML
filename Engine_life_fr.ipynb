## Contexte

Les moteurs commencent avec des niveaux variables d'usure initiale et des différences de fabrication, qui sont inconnus de l'utilisateur. Initialement, les moteurs fonctionnent dans des conditions normales et commencent à se dégrader au fil du temps. Cette dégradation augmente progressivement jusqu'à ce qu'un seuil prédéfini soit atteint, point auquel le moteur est considéré comme dangereux pour une utilisation ultérieure.

On vous donne des données multivariées provenant de moteurs d'avion du même type, enregistrées lors de cycles opérationnels aléatoires. Le jeu de données comprend trois paramètres de *réglage* (setting), 21 lectures de capteurs, et la *durée de vie* correspondante à cet état. La *durée de vie*, obtenue a posteriori, représente le nombre de cycles opérationnels restants avant que le seuil de dégradation ne soit atteint.

L'objectif est de prédire la durée de vie de ce type de moteur en fonction des paramètres de réglage et des lectures des capteurs à un état opérationnel donné, en utilisant des modèles de régression linéaire.

Les données sont divisées en un jeu de données d'entraînement (`train_data.dat`) et un jeu de données de test (`test_data.dat`). Le jeu de données de test est exclusivement réservé à l'évaluation des performances du modèle entraîné sur des données invisibles, et l'entraînement du modèle doit être effectué exclusivement à l'aide du jeu de données d'entraînement.
